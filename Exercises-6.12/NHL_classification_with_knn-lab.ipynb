### 8. Create a 50-50 train-test-split of your target and predictors. Refit the KNN and assess the accuracy.

```python
# A:

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.5,random_state=42,stratify=y)

knn = KNeighborsClassifier(n_neighbors=1)
knn.fit(X_train, y_train)
```

```python
train_accuracy = knn.score(X_train, y_train)
test_accuracy = knn.score(X_test, y_test)

print("Train Accuracy:", train_accuracy)
print("Test Accuracy:", test_accuracy)
```

### 9. Evaluate the test accuracy of a KNN where K == number of rows in the training data.

```python
# A:


knn = KNeighborsClassifier(n_neighbors=len(X_train))

knn.fit(X_train, y_train)

test_accuracy = knn.score(X_test, y_test)

print("K =", len(X_train))
print("Test Accuracy:", test_accuracy)
```

### 10. Fit the KNN at values of K from 1 to the number of rows in the training data.
- Store the test accuracy in a list.
- Plot the test accuracy vs. the number of neighbors.

```python
# A:
test_accuracy = []

k_values = range(1, len(X_train) + 1)

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    test_accuracy.append(knn.score(X_test, y_test))
```

```python
plt.figure(figsize=(8,5))

plt.plot(k_values, test_accuracy, marker='o')

plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy vs K')

plt.grid(True)

plt.show()
```

```python
best_k = test_accuracy.index(max(test_accuracy)) + 1

print("Best K:", best_k)
print("Best Accuracy:", max(test_accuracy))
```

### 11. Fit KNN across different values of K and plot the mean cross-validated accuracy with 5 folds.

```python
# A:
from sklearn.model_selection import cross_val_score
k_values = range(1, 46)
mean_cv_accuracy = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    
    cv_scores = cross_val_score(
        knn,
        X,
        y,
        cv=5,
        scoring='accuracy'
    )
    
    mean_cv_accuracy.append(cv_scores.mean())
```

```python
plt.figure(figsize=(8, 5))

plt.plot(k_values, mean_cv_accuracy, marker='o')

plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Mean Cross-Validated Accuracy')
plt.title('5-Fold Cross-Validated Accuracy vs K')
plt.grid(True)

plt.show()
```

```python
best_k = k_values[mean_cv_accuracy.index(max(mean_cv_accuracy))]
best_cv_accuracy = max(mean_cv_accuracy)

print("Best K:", best_k)
print("Best Mean CV Accuracy:", best_cv_accuracy)
```

### 12. Standardize the predictor matrix and cross-validate across the different K.
- Plot the standardized mean cross-validated accuracy against the unstandardized. Which is better?
- Why?

```python
# A:

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python


scaled_cv_accuracy = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)

    cv_scores = cross_val_score(
        knn,
        X_scaled,
        y,
        cv=5,
        scoring='accuracy'
    )

    scaled_cv_accuracy.append(cv_scores.mean())
```

```python

plt.figure(figsize=(8,5))

plt.plot(k_values, mean_cv_accuracy, marker='o', label='Unstandardized')
plt.plot(k_values, scaled_cv_accuracy, marker='s', label='Standardized')

plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Mean Cross-Validated Accuracy')
plt.title('Standardized vs Unstandardized KNN')
plt.legend()

plt.grid(True)
plt.show()
```

```python
print("Best Unstandardized Accuracy:", max(mean_cv_accuracy))
print("Best Standardized Accuracy:", max(scaled_cv_accuracy))
```

>Both standardized and unstandardized KNN achieved the same best mean cross-validated accuracy of 61.11%. Therefore, neither model is clearly better based on the maximum accuracy. However, standardization is generally preferred for KNN because it is a distance-based algorithm. Scaling ensures that all predictor variables contribute equally to the distance calculation and prevents features with larger numerical ranges from dominating.



## What was the best 'k'?

Using `GridSearch` & `Pipeline` find the best `k` for the KNN model

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier())
])

param_grid = {
    'knn__n_neighbors': range(1, 46)
}

grid = GridSearchCV(
    pipe,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)

grid.fit(X, y)
```

```python
print("Best K:", grid.best_params_['knn__n_neighbors'])
print("Best CV Accuracy:", grid.best_score_)
```

```python
best_model = grid.best_estimator_

print(best_model)
```

>Using GridSearchCV with a Pipeline, the best value of K is 3, with a mean cross-validated accuracy of 61.11%. The pipeline standardizes the data before fitting the KNN model, ensuring proper preprocessing during cross-validation.
